```python
from datasets import load_dataset

ds = load_dataset("ada-datadruids/full_tmdb_movies_dataset")
```

```python
import pandas as pd

df = ds["train"].to_pandas()
df
```

```python
df.info()
```

```python
df['overview'] = df['overview'].fillna('')
df = df.dropna(subset=['title'])
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('intfloat/multilingual-e5-small', device='cuda')
```

```python
sinopsis_list = df['overview'].tolist()
embeddings = model.encode(sinopsis_list, show_progress_bar=True, device='cuda')
```

```python
import numpy as np

np.save('movie_embeddings_XLM-RoBERTa (Small)', embeddings)
```
