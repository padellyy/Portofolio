```python
import numpy as np

emb_mbert = np.load("movie_embeddings_mBERT.npy")
emb_xlm = np.load("movie_embeddings_XLM-RoBERTa (Small).npy")
print(emb_mbert.shape[0] == emb_xlm.shape[0])
```

```python
hybrid_embeddings = np.hstack((emb_mbert, emb_xlm))
np.save("movie_embeddings_Hybrid_mBERT_XLM.npy", hybrid_embeddings)
```
